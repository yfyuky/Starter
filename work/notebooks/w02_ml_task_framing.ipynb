## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This task is a **classification** problem.

We aim to predict whether a customer will 'churn' (i.e., cancel their service or stop making purchases) within a certain future period (e.g., next month). This is a binary classification task, as the outcome variable has two distinct categories: 'churn' or 'not churn'.

This is chosen because:
*   **Clear Categories:** The target variable is categorical (churn/no churn), which is the defining characteristic of classification.
*   **Actionable Insights:** Predicting churn allows the business to proactively intervene with at-risk customers, offering incentives or support to retain them.
*   **Common Business Problem:** Customer churn is a widely recognized and impactful problem across many industries, making it a good candidate for demonstrating an ML task framing.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

We would predict the **target variable 'Churn'**.

This label comes from an **observed outcome**, specifically, a customer's actual behavior within a defined future period. For example:

*   **Observed Churn Event:** If a customer explicitly cancels their subscription or account within the next 30, 60, or 90 days.

The choice of the observation window (e.g., next 30 days) and the specific definition of 'churn' (explicit cancellation vs. inactivity) would be determined in collaboration with business stakeholders to align with their strategic goals and the operational actions they can take to retain customers.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric for the churn prediction model would be the **F1-score**.

**Why F1-score?**

1.  **Imbalanced Classes:** In customer churn prediction, the 'churn' class (customers who churn) is typically much smaller than the 'no churn' class (customers who stay). Accuracy alone can be misleading in such cases, as a model that predicts 'no churn' for all customers might still achieve high accuracy if churn is rare. F1-score, being the harmonic mean of precision and recall, is more robust to class imbalance.
2.  **Balancing Precision and Recall:**
    *   **Precision** focuses on the accuracy of positive predictions (of all customers predicted to churn, how many actually churned?). High precision means fewer false positives (i.e., not incorrectly targeting loyal customers with retention offers).
    *   **Recall** focuses on the model's ability to find all positive instances (of all customers who actually churned, how many did the model correctly identify?). High recall means fewer false negatives (i.e., not missing many customers who are about to churn).

    The business goal often involves both: identifying a good portion of actual churners (high recall) without irritating too many loyal customers (high precision). F1-score provides a single metric that balances these two competing concerns.

**What number means 'good'?**

A 'good' F1-score would depend on the business context and baseline performance. However, for a first iteration, an F1-score significantly higher than what a random classifier or simple heuristic would achieve would be considered a success. For example, an F1-score above **0.70** (on a held-out test set) could be a good initial target, indicating a reasonable balance between identifying churners and avoiding false alarms.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

For customer churn prediction, the **unit of analysis is one customer at a specific point in time**. Each row in our dataset would represent a snapshot of a single customer's profile and behavior up to a certain cutoff date, along with a label indicating whether that customer churned in the subsequent period.

**One row = One customer snapshot**

This snapshot would include:

*   **Customer Identifier:** A unique ID for the customer.
*   **Features:** Various attributes and behavioral metrics leading up to the prediction period (e.g., demographics, usage patterns, customer service interactions, billing history, tenure).
*   **Target Variable:** A binary flag (e.g., 0 or 1) indicating whether the customer churned within the defined future period (e.g., next 30 days) *after* the snapshot was taken.

This structure allows the model to learn patterns from past customer behavior and characteristics to predict future churn likelihood.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Customer churn is a complex phenomenon influenced by a multitude of interacting factors, making it exceptionally difficult to capture accurately with a simple, fixed rule (e.g., an `if-statement`). Here's why ML is superior:

1.  **Multivariate Interactions:** Churn is rarely due to a single cause. It's often a combination of factors like declining usage, multiple customer service complaints, recent price changes, tenure, and competitor offers. A fixed rule would struggle to weigh these factors and their complex, non-linear interactions simultaneously. For example, a low usage might be fine for a new customer but a strong churn signal for a long-term customer.
2.  **Non-linear Relationships:** The relationship between customer attributes and churn is rarely linear. A fixed threshold for a single variable (e.g., 'if usage < 10 then churn') might miss many nuances. ML models can learn non-linear patterns and complex decision boundaries that humans might not explicitly recognize or articulate.
3.  **Adaptability and Evolution:** Customer behavior and market dynamics are constantly changing. A fixed rule, once set, is static and would quickly become outdated. An ML model, however, can be continuously retrained with fresh data, allowing it to adapt to evolving patterns of churn and maintain its predictive power.
4.  **Hidden Patterns and Signals:** There might be subtle patterns in the data that are not immediately obvious to human analysts but are critical for predicting churn. ML algorithms excel at discovering these latent relationships across a vast number of features.
5.  **Quantifying Risk:** A fixed rule provides a binary outcome (churn/no churn). ML models, especially classification models, often provide a probability score for churn. This probability allows for more granular risk assessment and targeted interventions (e.g., targeting high-risk customers with specific retention campaigns).
