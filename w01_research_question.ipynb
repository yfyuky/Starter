## 2. The question: decision, action, cost of a wrong call

**Research question:** Which pages should be reviewed first for refresh actions, based on observable signals of risk/opportunity?

- **Decision improved:** Prioritization of pages for manual review.
- **Primary actor:** Content/SEO reviewer (or content operations team).
- **Unit of analysis (grain):** One row per content page (starter dataset row).
- **Output:** Ranked queue of pages with score + reason codes (e.g., stale + visible, low CTR at meaningful position, decline risk).
- **Action from output:** Review/update title/meta, expand/thin-content fixes, refresh outdated sections, or monitor instead of edit.
- **Cost of wrong recommendation:**
  - **False positive:** Team spends time reviewing a page with low payoff.
  - **False negative:** A high-impact page is missed and continues to underperform.
- **Why data/ML helps:** Manual prioritization is noisy across thousands of pages; a transparent ranking can sort by likely impact and make limited review capacity more effective.

## 3. Quick look at the data (2-3 real numbers)

First, let's download the necessary data file.

```python
import os

# Create the 'data/raw' directory if it doesn't exist
os.makedirs('data/raw', exist_ok=True)

# Download the file from the GitHub repository
!wget -q -O data/raw/content_refresh_anonymized.csv https://github.com/yfyuky/Starter/raw/main/data/raw/content_refresh_anonymized.csv
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Label used in starter exercises
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Hand rule from Week 2 idea
stale = (df["days_since_last_update"] >= 180).astype(int)
visible = (df["impressions_90d"] >= 500).astype(int)
df["hand_rule_score"] = stale * visible * df["impressions_90d"]

def precision_at_k(scores, labels, k):
    order = np.argsort(-np.asarray(scores))
    topk = np.asarray(labels)[order[:k]]
    return topk.mean()

n_pages = len(df)
declining_rate = df["is_declining_label"].mean()
stale_visible_count = int(((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum())
p_at_50 = precision_at_k(df["hand_rule_score"], df["is_declining_label"].values, 50)

print(f"Total pages: {n_pages}")
print(f"Declining-label rate: {declining_rate:.3f}")
print(f"Stale+visible pages (>=180 days since update & >=500 impressions): {stale_visible_count}")
print(f"Hand-rule Precision@50: {p_at_50:.3f}")
```

## 4. Careful words: what I can and can't claim

This project is **decision-support**, not causal proof.  
I can claim observed, directional findings such as: “pages with this signal pattern are more likely to appear in high-priority review candidates under this label definition.” I can evaluate ranking quality with metrics like Precision@K and compare model vs baseline rules.

I **cannot** claim:  
- that a refresh action will definitely cause recovery,  
- that I proved search algorithm factors,  
- or that this model is universally correct outside the defined data window and validation design.

So the output should be interpreted as a **prioritization aid** for human reviewers, not an automatic truth engine.
